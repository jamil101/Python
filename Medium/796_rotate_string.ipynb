**796. Rotate String**

**Easy**

**Companies**:Amazon Apple eBay LinkedIn Microsoft

Given two strings s and goal, return true if and only if s can become goal after some number of shifts on s.

A shift on s consists of moving the leftmost character of s to the rightmost position.

For example, if s = "abcde", then it will be "bcdea" after one shift.

**Example 1:**

```python
Input: s = "abcde", goal = "cdeab"
Output: true
```

**Example 2:**

```python
Input: s = "abcde", goal = "abced"
Output: false
```

**Constraints:**

- 1 <= s.length, goal.length <= 100
- s and goal consist of lowercase English letters.


````python
class Solution:
    def rotateString(self, s: str, goal: str) -> bool:
        """
        Determines if string s can be rotated to become goal.

        Algorithm:
            - If lengths differ, return False.
            - Concatenate s with itself (s + s).
            - Check if goal is a substring of (s + s).

        Args:
            s (str): Original string.
            goal (str): Target string after rotations.

        Returns:
            bool: True if goal is a rotation of s, else False.

        Time Complexity:
            O(n) — substring search.

        Space Complexity:
            O(n) — for concatenated string.
        """
        if len(s) != len(goal):
            return False
        return goal in (s + s)
````

````python

class Solution:
    def rotateString(self, s: str, goal: str) -> bool:
        """
        Checks all possible rotations of s.

        Algorithm:
            - If lengths differ, return False.
            - Generate all rotations using slicing.
            - Compare each rotation with goal.

        Args:
            s (str): Original string.
            goal (str): Target string.

        Returns:
            bool: True if any rotation matches goal.

        Time Complexity:
            O(n^2) — n rotations * O(n) comparison.

        Space Complexity:
            O(n) — temporary rotated string.
        """
        if len(s) != len(goal):
            return False

        for i in range(len(s)):
            rotated = s[i:] + s[:i]
            if rotated == goal:
                return True

        return False
````

````python
from collections import deque

class Solution:
    def rotateString(self, s: str, goal: str) -> bool:
        """
        Uses deque to simulate rotation.

        Algorithm:
            - Convert string to deque.
            - Perform left rotation using popleft() and append().
            - Compare after each rotation.

        Args:
            s (str): Original string.
            goal (str): Target string.

        Returns:
            bool: True if rotation matches goal.

        Time Complexity:
            O(n^2) — each join operation costs O(n).

        Space Complexity:
            O(n) — deque storage.
        """
        if len(s) != len(goal):
            return False

        dq = deque(s)

        for _ in range(len(s)):
            if ''.join(dq) == goal:
                return True
            dq.append(dq.popleft())

        return False
````
