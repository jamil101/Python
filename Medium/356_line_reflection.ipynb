**356. Line Reflection**

**Medium**

**Company: Amazon Google**

Given n points on a 2D plane, find if there is such a line parallel to y-axis that reflect the given points.

**Example 1:**
```python
Input: [[1,1],[-1,1]]
Output: true
```
**Example 2:**
```python
Input: [[1,1],[-1,-1]]
Output: false
```
**Constraints:**

- n == points.length 
- 1 <= n <= 10^4
- -10^8 <= points[i][j] <= 10^8

````python
class Solution:
    def isReflected(self, points):

        """
        ---------------------------------------------------------
        Algorithm:

        1. Find minimum and maximum x-coordinate.

        2. Compute:

               reflection_sum = minX + maxX

           If reflection line exists:

               x = c

           then every reflected pair satisfies:

               x1 + x2 = 2 * c

        3. For every point (x, y):

               reflected point =
               (reflection_sum - x, y)

           Search entire array to check whether
           reflected point exists.

        4. If any reflected point is missing:

               return False

        5. Otherwise return True.

        ---------------------------------------------------------

        Time Complexity:
            O(n^2)

        Space Complexity:
            O(1)

        ---------------------------------------------------------
        """

        minX = float('inf')
        maxX = float('-inf')

        # Find leftmost and rightmost x-coordinate
        for x, y in points:
            minX = min(minX, x)
            maxX = max(maxX, x)

        reflection_sum = minX + maxX

        # Check reflection for every point
        for x, y in points:

            reflected_x = reflection_sum - x
            found = False

            # Brute force search
            for a, b in points:

                if a == reflected_x and b == y:
                    found = True
                    break

            if not found:
                return False

        return True
````

````python
class Solution:
    def isReflected(self, points):

        """
        ---------------------------------------------------------
        Algorithm:

        1. Find minimum and maximum x-coordinate.

        2. Compute:

               reflection_sum = minX + maxX

        3. Store every point inside a hash set
           for O(1) lookup.

        4. For every point (x, y):

               reflected point =
               (reflection_sum - x, y)

           Check whether reflected point exists
           in the hash set.

        5. If any reflected point is missing:

               return False

        6. Otherwise return True.

        ---------------------------------------------------------

        Time Complexity:
            O(n)

        Space Complexity:
            O(n)

        ---------------------------------------------------------
        """

        minX = float('inf')
        maxX = float('-inf')

        point_set = set()

        # Store points and find min/max x
        for x, y in points:

            minX = min(minX, x)
            maxX = max(maxX, x)

            point_set.add((x, y))

        reflection_sum = minX + maxX

        # Verify reflection for every point
        for x, y in points:

            reflected_point = (reflection_sum - x, y)

            if reflected_point not in point_set:
                return False

        return True
````

````python
from collections import defaultdict

class Solution:
    def isReflected(self, points):

        """
        ---------------------------------------------------------
        Algorithm:

        1. Group all points according to y-coordinate.

        2. Find:

               reflection_sum = minX + maxX

        3. For every horizontal row:

               a) Sort x-coordinates

               b) Use two pointers:

                      left  -> smallest x
                      right -> largest x

               c) Symmetric pairs must satisfy:

                      xs[left] + xs[right]
                      == reflection_sum

        4. If any row violates symmetry:

               return False

        5. Otherwise return True.

        ---------------------------------------------------------

        Time Complexity:
            O(n log n)

        Space Complexity:
            O(n)

        ---------------------------------------------------------
        """

        rows = defaultdict(list)

        minX = float('inf')
        maxX = float('-inf')

        # Group points by y-coordinate
        for x, y in points:

            rows[y].append(x)

            minX = min(minX, x)
            maxX = max(maxX, x)

        reflection_sum = minX + maxX

        # Check symmetry for every row
        for y in rows:

            xs = sorted(rows[y])

            left = 0
            right = len(xs) - 1

            while left <= right:

                if xs[left] + xs[right] != reflection_sum:
                    return False

                left += 1
                right -= 1

        return True
````
