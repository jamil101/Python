**1838. Frequency of the Most Frequent Element**

The frequency of an element is the number of times it occurs in an array.

You are given an integer array nums and an integer k. In one operation, you can choose an index of nums and increment the element at that index by 1.

Return the maximum possible frequency of an element after performing at most k operations.

**Example 1:**
```python
Input: nums = [1,2,4], k = 5
Output: 3
```
**Explanation:** Increment the first element three times and the second element two times to make nums = [4,4,4].
4 has a frequency of 3.

**Example 2:**
```python
Input: nums = [1,4,8,13], k = 5
Output: 2
```
**Explanation:** There are multiple optimal solutions:
- Increment the first element three times to make nums = [4,4,8,13]. 4 has a frequency of 2.
- Increment the second element four times to make nums = [1,8,8,13]. 8 has a frequency of 2.
- Increment the third element five times to make nums = [1,4,13,13]. 13 has a frequency of 2.

**Example 3:**
```python
Input: nums = [3,9,6], k = 2
Output: 1
 ```

**Constraints:**

- 1 <= nums.length <= 10<sup>5</sup>
- 1 <= nums[i] <= 10<sup>5</sup>
- 1 <= k <= 10<sup>5</sup>

````python
import bisect

class Solution:
    def maxFrequency(self, nums, k):
        """
        Algorithm:
        1. Sort nums
        2. Build prefix sum array
        3. For each index i:
           - Binary search left boundary
           - Check if we can make all elements equal to nums[i]

        Cost formula:
        cost = nums[i] * length - subarray_sum

        Time Complexity: O(n log n)
        Space Complexity: O(n)
        """

        nums.sort()
        n = len(nums)

        prefix = [0] * (n + 1)
        for i in range(n):
            prefix[i+1] = prefix[i] + nums[i]

        max_freq = 1

        for i in range(n):
            left, right = 0, i

            while left <= right:
                mid = (left + right) // 2
                length = i - mid + 1
                sub_sum = prefix[i+1] - prefix[mid]

                cost = nums[i] * length - sub_sum

                if cost <= k:
                    max_freq = max(max_freq, length)
                    right = mid - 1
                else:
                    left = mid + 1

        return max_freq
````

````python
class Solution:
    def maxFrequency(self, nums, k):
        """
        Algorithm (Sliding Window):
        1. Sort nums
        2. Use two pointers: left, right
        3. Maintain current window sum
        4. Goal:
           Make all elements in window equal to nums[right]

        Cost formula:
        cost = nums[right] * window_size - window_sum

        5. If cost > k:
           shrink window from left

        6. Track maximum window size

        Time Complexity: O(n log n) (sorting) + O(n)
        Space Complexity: O(1)
        """

        nums.sort()

        left = 0
        total = 0
        max_freq = 1

        for right in range(len(nums)):
            total += nums[right]

            # Check if cost exceeds k
            while nums[right] * (right - left + 1) - total > k:
                total -= nums[left]
                left += 1

            max_freq = max(max_freq, right - left + 1)

        return max_freq
````
