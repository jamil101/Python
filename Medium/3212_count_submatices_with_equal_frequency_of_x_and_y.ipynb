**3212. Count Submatrices With Equal Frequency of X and Y**

**Medium**

**Companies**

Given a 2D character matrix grid, where grid[i][j] is either 'X', 'Y', or '.', return the number of submatrices that contain:

- grid[0][0]
- an equal frequency of 'X' and 'Y'.
- at least one 'X'.
 

**Example 1:**
```python
Input: grid = [["X","Y","."],["Y",".","."]]
Output: 3
```

**Example 2:**
```python
Input: grid = [["X","X"],["X","Y"]]
Output: 0
```
**Explanation:**

No submatrix has an equal frequency of 'X' and 'Y'.

**Example 3:**
```python
Input: grid = [[".","."],[".","."]]
Output: 0
```
**Explanation:**
No submatrix has at least one 'X'.

**Constraints:**

- 1 <= grid.length, grid[i].length <= 1000
- grid[i][j] is either 'X', 'Y', or '.'.

````python
class Solution:
    def numberOfSubmatrices(self, grid):
        """
        Algorithm:
        1. Create two prefix matrices:
           px[i][j] = count of 'X' from (0,0) to (i,j)
           py[i][j] = count of 'Y' from (0,0) to (i,j)

        2. Use prefix sum formula:
           px[i][j] = current + top + left - top-left
           same for py

        3. For each (i, j):
           - If px[i][j] == py[i][j] AND px[i][j] > 0:
             increment count

        4. Return count

        Time Complexity: O(m * n)
        Space Complexity: O(m * n)
        """

        m, n = len(grid), len(grid[0])

        px = [[0] * n for _ in range(m)]
        py = [[0] * n for _ in range(m)]

        count = 0

        for i in range(m):
            for j in range(n):

                # Current cell contribution
                px[i][j] = 1 if grid[i][j] == 'X' else 0
                py[i][j] = 1 if grid[i][j] == 'Y' else 0

                if i > 0:
                    px[i][j] += px[i-1][j]
                    py[i][j] += py[i-1][j]

                if j > 0:
                    px[i][j] += px[i][j-1]
                    py[i][j] += py[i][j-1]

                if i > 0 and j > 0:
                    px[i][j] -= px[i-1][j-1]
                    py[i][j] -= py[i-1][j-1]

                # Check conditions
                if px[i][j] == py[i][j] and px[i][j] > 0:
                    count += 1

        return count
````

````python
class Solution:
    def numberOfSubmatrices(self, grid):
        """
        Algorithm:
        1. Convert grid:
           X -> +1
           Y -> -1
           . -> 0

        2. Build prefix sum matrix
        3. For each (i,j):
           - If sum == 0 → equal X and Y
           - Also check at least one X using separate count

        Time Complexity: O(m * n)
        Space Complexity: O(m * n)
        """

        m, n = len(grid), len(grid[0])

        pref = [[0]*n for _ in range(m)]
        countX = [[0]*n for _ in range(m)]

        ans = 0

        for i in range(m):
            for j in range(n):

                val = 0
                if grid[i][j] == 'X':
                    val = 1
                elif grid[i][j] == 'Y':
                    val = -1

                pref[i][j] = val
                countX[i][j] = 1 if grid[i][j] == 'X' else 0

                if i > 0:
                    pref[i][j] += pref[i-1][j]
                    countX[i][j] += countX[i-1][j]

                if j > 0:
                    pref[i][j] += pref[i][j-1]
                    countX[i][j] += countX[i][j-1]

                if i > 0 and j > 0:
                    pref[i][j] -= pref[i-1][j-1]
                    countX[i][j] -= countX[i-1][j-1]

                if pref[i][j] == 0 and countX[i][j] > 0:
                    ans += 1

        return ans
````
