**581. Shortest Unsorted Continuous Subarray**

**Medium**

**Companies**:  Amazon Bloomberg Google LiveRamp Redfin Uber

Given an integer array nums, you need to find one continuous subarray such that if you only sort this subarray in non-decreasing order, then the whole array will be sorted in non-decreasing order.

Return the shortest such subarray and output its length.

**Example 1:**
```python
Input: nums = [2,6,4,8,10,9,15]
Output: 5
```
**Explanation:** You need to sort [6, 4, 8, 10, 9] in ascending order to make the whole array sorted in ascending order.

**Example 2:**
```python
Input: nums = [1,2,3,4]
Output: 0
```
**Example 3:**
```python
Input: nums = [1]
Output: 0
```
**Constraints:**

- 1 <= nums.length <= 104
- -10<sup>5 </sup><= nums[i] <= 10<sup>5</sup>
 

**Follow up:** Can you solve it in O(n) time complexity?

````python
class Solution:
    def findUnsortedSubarray(self, nums):
        """
        Algorithm:
        1. Create a sorted copy of array
        2. Compare original and sorted
        3. Find:
           - leftmost mismatch
           - rightmost mismatch
        4. Length = right - left + 1

        Time Complexity: O(n log n)
        Space Complexity: O(n)
        """

        sorted_nums = sorted(nums)

        left = 0
        right = len(nums) - 1

        while left < len(nums) and nums[left] == sorted_nums[left]:
            left += 1

        while right > left and nums[right] == sorted_nums[right]:
            right -= 1

        return 0 if left >= right else right - left + 1
````

````python
class Solution:
    def findUnsortedSubarray(self, nums):
        """
        Algorithm:
        1. Traverse from left → right:
           - Track max_so_far
           - If nums[i] < max_so_far → update right boundary

        2. Traverse from right → left:
           - Track min_so_far
           - If nums[i] > min_so_far → update left boundary

        3. If no violation → return 0
        4. Else return right - left + 1

        Time Complexity: O(n)
        Space Complexity: O(1)
        """

        n = len(nums)

        left, right = -1, -2  # ensures 0 if already sorted
        max_so_far = nums[0]
        min_so_far = nums[-1]

        # Left → Right
        for i in range(1, n):
            max_so_far = max(max_so_far, nums[i])
            if nums[i] < max_so_far:
                right = i

        # Right → Left
        for i in range(n-2, -1, -1):
            min_so_far = min(min_so_far, nums[i])
            if nums[i] > min_so_far:
                left = i

        return right - left + 1
````

````python
class Solution:
    def findUnsortedSubarray(self, nums):
        """
        Algorithm:
        1. Use increasing stack to find left boundary
        2. Use decreasing stack to find right boundary

        Time Complexity: O(n)
        Space Complexity: O(n)
        """

        n = len(nums)
        stack = []
        left = n
        right = 0

        # Find left boundary
        for i in range(n):
            while stack and nums[stack[-1]] > nums[i]:
                left = min(left, stack.pop())
            stack.append(i)

        stack = []

        # Find right boundary
        for i in range(n-1, -1, -1):
            while stack and nums[stack[-1]] < nums[i]:
                right = max(right, stack.pop())
            stack.append(i)

        return right - left + 1 if right > left else 0
````
