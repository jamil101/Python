**3070. Count Submatrices with Top-Left Element and Sum Less Than k**

**Medium**

**Companies**
You are given a 0-indexed integer matrix grid and an integer k.

Return the number of submatrices that contain the top-left element of the grid, and have a sum less than or equal to k.

 

**Example 1:**
```python
Input: grid = [[7,6,3],[6,6,1]], k = 18
Output: 4
```
**Explanation:** There are only 4 submatrices, shown in the image above, that contain the top-left element of grid, and have a sum less than or equal to 18.

**Example 2:**
```python
Input: grid = [[7,2,9],[1,5,0],[2,6,6]], k = 20
Output: 6
```
**Explanation:** There are only 6 submatrices, shown in the image above, that contain the top-left element of grid, and have a sum less than or equal to 20.
 
**Constraints:**

- m == grid.length 
- n == grid[i].length
- 1 <= n, m <= 1000 
- 0 <= grid[i][j] <= 1000
- 1 <= k <= 109

````python
class Solution:
    def countSubmatrices(self, grid, k):
        """
        Algorithm:
        1. Create a prefix sum matrix `pref`
        2. pref[i][j] stores sum of submatrix (0,0) → (i,j)
        3. Use formula:
           pref[i][j] = grid[i][j]
                        + pref[i-1][j]
                        + pref[i][j-1]
                        - pref[i-1][j-1]
        4. For each cell (i, j):
           - Check if pref[i][j] <= k
           - If yes, increment count
        5. Return count

        Time Complexity: O(m * n)
        Space Complexity: O(m * n)
        """

        m, n = len(grid), len(grid[0])
        pref = [[0] * n for _ in range(m)]
        count = 0

        for i in range(m):
            for j in range(n):
                pref[i][j] = grid[i][j]

                if i > 0:
                    pref[i][j] += pref[i-1][j]
                if j > 0:
                    pref[i][j] += pref[i][j-1]
                if i > 0 and j > 0:
                    pref[i][j] -= pref[i-1][j-1]

                if pref[i][j] <= k:
                    count += 1

        return count
````

````python
class Solution:
    def countSubmatrices(self, grid, k):
        """
        Algorithm:
        1. Instead of using extra matrix, reuse `grid` as prefix sum
        2. Update each cell to store sum from (0,0) → (i,j)
        3. Same formula as prefix sum
        4. Check condition during update

        Time Complexity: O(m * n)
        Space Complexity: O(1)
        """

        m, n = len(grid), len(grid[0])
        count = 0

        for i in range(m):
            for j in range(n):

                if i > 0:
                    grid[i][j] += grid[i-1][j]
                if j > 0:
                    grid[i][j] += grid[i][j-1]
                if i > 0 and j > 0:
                    grid[i][j] -= grid[i-1][j-1]

                if grid[i][j] <= k:
                    count += 1

        return count
````
