**159. Longest Substring with At Most Two Distinct Characters**

**Medium**

**Companies:** Adobe Amazon Facebook Google Microsoft

Given a string s , find the length of the longest substring t  that contains at most 2 distinct characters.

**Example 1:**
```python
Input: "eceba"
Output: 3
```
**Explanation:** t is "ece" which its length is 3.
**Example 2:**
```python
Input: "ccaabbb"
Output: 5
```
**Explanation:** t is "aabbb" which its length is 5.


**Constraints:**

- 1 <= s.length <= 10^5
- s consists of English letters.

````python
class Solution:
    def lengthOfLongestSubstringTwoDistinct(self, s: str) -> int:
        """
        Algorithm:
        1. Generate all substrings
        2. For each substring:
           - Count distinct characters using set
        3. If distinct <= 2:
           - Update max length

        Time Complexity: O(n^3)
        Space Complexity: O(n)
        """

        n = len(s)
        max_len = 0

        for i in range(n):
            for j in range(i, n):
                substring = s[i:j+1]
                if len(set(substring)) <= 2:
                    max_len = max(max_len, j - i + 1)

        return max_len
````

````python
class Solution:
    def lengthOfLongestSubstringTwoDistinct(self, s: str) -> int:
        """
        Algorithm (Sliding Window):
        1. Use two pointers: left, right
        2. Expand right pointer
        3. Maintain frequency map of characters
        4. If distinct characters > 2:
           - Shrink window from left
           - Remove characters with freq 0
        5. Update max length at each step

        Time Complexity: O(n)
        Space Complexity: O(1)
        """

        left = 0
        freq = {}
        max_len = 0

        for right in range(len(s)):
            # Add current character
            freq[s[right]] = freq.get(s[right], 0) + 1

            # Shrink window if more than 2 distinct characters
            while len(freq) > 2:
                freq[s[left]] -= 1
                if freq[s[left]] == 0:
                    del freq[s[left]]
                left += 1

            # Update result
            max_len = max(max_len, right - left + 1)

        return max_len
````

````python
class Solution:
    def lengthOfLongestSubstringTwoDistinct(self, s: str) -> int:
        """
        Algorithm:
        1. Use dictionary to store last index of each character
        2. Expand right pointer
        3. If size > 2:
           - Remove character with smallest index
           - Move left pointer

        Time Complexity: O(n)
        Space Complexity: O(1)
        """

        left = 0
        last_index = {}
        max_len = 0

        for right, ch in enumerate(s):
            last_index[ch] = right

            if len(last_index) > 2:
                # Find character with smallest last index
                del_idx = min(last_index.values())
                
                # Remove that character
                for key in list(last_index.keys()):
                    if last_index[key] == del_idx:
                        del last_index[key]
                        break

                # Move left pointer
                left = del_idx + 1

            max_len = max(max_len, right - left + 1)

        return max_len
````

````python
class Solution:
    def lengthOfLongestSubstringTwoDistinct(self, s: str) -> int:
        """
        Algorithm:
        1. Track:
           - last char
           - second last char
           - current window length
        2. Extend window if valid
        3. Reset when third char appears

        Time Complexity: O(n)
        Space Complexity: O(1)
        """

        if len(s) < 3:
            return len(s)

        last_char = ''
        second_last_char = ''
        last_char_count = 0
        curr_len = 0
        max_len = 0

        for ch in s:
            if ch == last_char or ch == second_last_char:
                curr_len += 1
            else:
                curr_len = last_char_count + 1

            if ch == last_char:
                last_char_count += 1
            else:
                last_char_count = 1
                second_last_char = last_char
                last_char = ch

            max_len = max(max_len, curr_len)

        return max_len
````
