**862. Shortest Subarray with Sum at Least K**

**Hard**

**Companies**: Amazon Facebook Goldman Sachs Google

Given an integer array nums and an integer k, return the length of the shortest non-empty subarray of nums with a sum of at least k. If there is no such subarray, return -1.

A subarray is a contiguous part of an array.

 

**Example 1:**
```python
Input: nums = [1], k = 1
Output: 1
```
**Example 2:**
```python
Input: nums = [1,2], k = 4
Output: -1
```
**Example 3:**
```python
Input: nums = [2,-1,2], k = 3
Output: 3
```
**Constraints:**

- 1 <= nums.length <= 105
- -105 <= nums[i] <= 105
- 1 <= k <= 109

````python
from collections import deque

class Solution:
    def shortestSubarray(self, nums, k):
        """
        Algorithm:
        1. Compute prefix sum array P
           P[i] = sum of nums[0...i-1]

        2. Use deque to store indices of prefix array

        3. For each index j:
           - While deque not empty AND
             P[j] - P[dq[0]] >= k:
                update answer
                pop left (we found smaller subarray)

           - Maintain monotonic increasing deque:
             While dq not empty AND
             P[j] <= P[dq[-1]]:
                pop right

           - Append current index j

        4. Return shortest length or -1

        Time Complexity: O(n)
        Space Complexity: O(n)
        """

        n = len(nums)

        # Step 1: Prefix sum
        prefix = [0] * (n + 1)
        for i in range(n):
            prefix[i+1] = prefix[i] + nums[i]

        dq = deque()
        min_len = float('inf')

        for j in range(n + 1):

            # Step 2: Check valid subarray
            while dq and prefix[j] - prefix[dq[0]] >= k:
                min_len = min(min_len, j - dq[0])
                dq.popleft()

            # Step 3: Maintain increasing order
            while dq and prefix[j] <= prefix[dq[-1]]:
                dq.pop()

            dq.append(j)

        return min_len if min_len != float('inf') else -1
````

````python
import heapq

class Solution:
    def shortestSubarray(self, nums, k):
        """
        Algorithm:
        1. Use prefix sum
        2. Use min-heap to track smallest prefix
        3. Check condition

        Time Complexity: O(n log n)
        Space Complexity: O(n)

        NOTE: Slower than deque solution
        """

        n = len(nums)
        prefix = 0
        heap = [(0, -1)]  # (prefix_sum, index)
        min_len = float('inf')

        for i in range(n):
            prefix += nums[i]

            while heap and prefix - heap[0][0] >= k:
                prev_sum, idx = heapq.heappop(heap)
                min_len = min(min_len, i - idx)

            heapq.heappush(heap, (prefix, i))

        return min_len if min_len != float('inf') else -1
````
