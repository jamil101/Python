**1886. Determine Whether Matrix Can Be Obtained By Rotation**

**Easy**

**Companies**

Given two n x n binary matrices mat and target, return true if it is possible to make mat equal to target by rotating mat in 90-degree increments, or false otherwise.

 

**Example 1:**
```python
Input: mat = [[0,1],[1,0]], target = [[1,0],[0,1]]
Output: true
```
**Explanation:** We can rotate mat 90 degrees clockwise to make mat equal target.

**Example 2:**
```python
Input: mat = [[0,1],[1,1]], target = [[1,0],[0,1]]
Output: false
```
**Explanation:** It is impossible to make mat equal to target by rotating mat.

**Example 3:**
```python
Input: mat = [[0,0,0],[0,1,0],[1,1,1]], target = [[1,1,1],[0,1,0],[0,0,0]]
Output: true
```
**Explanation:** We can rotate mat 90 degrees clockwise two times to make mat equal target.
 
**Constraints:**

- n == mat.length == target.length
- n == mat[i].length == target[i].length
- 1 <= n <= 10
- mat[i][j] and target[i][j] are either 0 or 1.

````python
class Solution:
    def findRotation(self, mat, target):
        """
        Algorithm:
        1. Try all 4 rotations:
           - 0°, 90°, 180°, 270°
        2. After each rotation:
           - Compare mat with target
        3. If any match → return True
        4. Else return False

        Time Complexity: O(4 * n^2)
        Space Complexity: O(1)
        """

        def rotate(matrix):
            # Rotate 90° clockwise
            n = len(matrix)
            new = [[0]*n for _ in range(n)]

            for i in range(n):
                for j in range(n):
                    new[j][n-1-i] = matrix[i][j]

            return new

        for _ in range(4):
            if mat == target:
                return True
            mat = rotate(mat)

        return False
````

````python
class Solution:
    def findRotation(self, mat, target):
        """
        Algorithm:
        1. Rotate matrix in-place using:
           transpose + reverse rows
        2. After each rotation, check equality
        3. Repeat 4 times

        Time Complexity: O(4 * n^2)
        Space Complexity: O(1)
        """

        def rotate_inplace(matrix):
            n = len(matrix)

            # Step 1: Transpose
            for i in range(n):
                for j in range(i+1, n):
                    matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]

            # Step 2: Reverse each row
            for row in matrix:
                row.reverse()

        for _ in range(4):
            if mat == target:
                return True
            rotate_inplace(mat)

        return False
````

````python
class Solution:
    def findRotation(self, mat, target):
        """
        Algorithm:
        1. Instead of rotating matrix physically,
           check positions mathematically

        Rotation mapping:
        90°:  mat[i][j] -> target[j][n-1-i]
        180°: mat[i][j] -> target[n-1-i][n-1-j]
        270°: mat[i][j] -> target[n-1-j][i]

        2. Check all 4 rotations in one pass

        Time Complexity: O(n^2)
        Space Complexity: O(1)
        """

        n = len(mat)

        same = True
        rot90 = True
        rot180 = True
        rot270 = True

        for i in range(n):
            for j in range(n):
                if mat[i][j] != target[i][j]:
                    same = False
                if mat[i][j] != target[j][n-1-i]:
                    rot90 = False
                if mat[i][j] != target[n-1-i][n-1-j]:
                    rot180 = False
                if mat[i][j] != target[n-1-j][i]:
                    rot270 = False

        return same or rot90 or rot180 or rot270
````

````python
class Solution:
    def findRotation(self, mat, target):
        """
        Algorithm:
        1. Use zip() and slicing to rotate
        2. Check after each rotation

        Time Complexity: O(4 * n^2)
        Space Complexity: O(n^2)
        """

        for _ in range(4):
            if mat == target:
                return True

            # Rotate using zip
            mat = [list(row) for row in zip(*mat[::-1])]

        return False
````
