**696. Count Binary Substrings**

**Easy**

**Companies**: Helix

Given a binary string s, return the number of non-empty substrings that have the same number of 0's and 1's, and all the 0's and all the 1's in these substrings are grouped consecutively.

Substrings that occur multiple times are counted the number of times they occur.

 

**Example 1:**
```python
Input: s = "00110011"
Output: 6
```
**Explanation:** There are 6 substrings that have equal number of consecutive 1's and 0's: "0011", "01", "1100", "10", "0011", and "01".
Notice that some of these substrings repeat and are counted the number of times they occur.

Also, "00110011" is not a valid substring because all the 0's (and 1's) are not grouped together.

**Example 2:**
```python
Input: s = "10101"
Output: 4
```
**Explanation**: There are 4 substrings: "10", "01", "10", "01" that have equal number of consecutive 1's and 0's.
 

**Constraints:**

- 1 <= s.length <= 105
- s[i] is either '0' or '1'.

````python
class Solution:
    def countBinarySubstrings(self, s: str) -> int:
        """
        Counts valid binary substrings using group counting.

        Algorithm:
        1. Traverse string and count lengths of consecutive groups.
           Example: "00110011" → [2,2,2,2]
        2. For each adjacent pair of groups:
           Add min(group[i], group[i+1]) to result.
        3. Return total.

        Time Complexity: O(n)
            - Single pass to build groups
            - Single pass to compute result

        Space Complexity: O(n)
            - Stores group lengths
        """
        groups = []
        count = 1

        # Step 1: Build consecutive group lengths
        for i in range(1, len(s)):
            if s[i] == s[i - 1]:
                count += 1
            else:
                groups.append(count)
                count = 1
        groups.append(count)

        # Step 2: Count valid substrings
        result = 0
        for i in range(len(groups) - 1):
            result += min(groups[i], groups[i + 1])

        return result
````

````python
class Solution:
    def countBinarySubstrings(self, s: str) -> int:
        """
        Counts valid binary substrings using optimized one-pass approach.

        Algorithm:
        1. Maintain:
            - prev_group_length
            - curr_group_length
        2. Traverse string:
            - If same char → increase curr_group_length
            - If different → add min(prev, curr) to result
              then update prev = curr, reset curr = 1
        3. After loop, add final min(prev, curr)

        Time Complexity: O(n)
            - Single pass

        Space Complexity: O(1)
            - Only few variables used
        """
        prev = 0
        curr = 1
        result = 0

        for i in range(1, len(s)):
            if s[i] == s[i - 1]:
                curr += 1
            else:
                result += min(prev, curr)
                prev = curr
                curr = 1

        # Add last pair
        result += min(prev, curr)

        return result
````
