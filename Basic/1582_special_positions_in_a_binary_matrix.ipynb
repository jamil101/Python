**1582. Special Positions in a Binary Matrix**

**Easy**

**Companies**: Google

Given an m x n binary matrix mat, return the number of special positions in mat.

A position (i, j) is called special if mat[i][j] == 1 and all other elements in row i and column j are 0 (rows and columns are 0-indexed).

**Example 1:**
```
Input: mat = [[1,0,0],[0,0,1],[1,0,0]]
Output: 1
```
**Explanation**: (1, 2) is a special position because mat[1][2] == 1 and all other elements in row 1 and column 2 are 0.

**Example 2:**
```
Input: mat = [[1,0,0],[0,1,0],[0,0,1]]
Output: 3
```
**Explanation**: (0, 0), (1, 1) and (2, 2) are special positions.
 
**Constraints:**

- m == mat.length
- n == mat[i].length
- 1 <= m, n <= 100
- mat[i][j] is either 0 or 1.

````python
class Solution:
    def numSpecial(self, mat: List[List[int]]) -> int:
        """
        Algorithm:
        1. For each cell (i, j):
             If mat[i][j] == 1:
                 - Count number of 1s in row i
                 - Count number of 1s in column j
                 - If both counts == 1, increment result
        2. Return result

        Time Complexity:
            O(m * n * (m + n))

        Space Complexity:
            O(1)
        """

        m, n = len(mat), len(mat[0])
        count = 0

        for i in range(m):
            for j in range(n):
                if mat[i][j] == 1:
                    row_count = sum(mat[i])
                    col_count = sum(mat[k][j] for k in range(m))

                    if row_count == 1 and col_count == 1:
                        count += 1

        return count
````

````python
class Solution:
    def numSpecial(self, mat: List[List[int]]) -> int:
        """
        Algorithm:
        1. Create row[] and col[] arrays
        2. First pass:
             Count number of 1s in each row and column
        3. Second pass:
             If mat[i][j] == 1 AND row[i] == 1 AND col[j] == 1
                 increment count
        4. Return count

        Time Complexity:
            O(m * n)

        Space Complexity:
            O(m + n)
        """

        m, n = len(mat), len(mat[0])
        row = [0] * m
        col = [0] * n

        # First pass
        for i in range(m):
            for j in range(n):
                if mat[i][j] == 1:
                    row[i] += 1
                    col[j] += 1

        count = 0

        # Second pass
        for i in range(m):
            for j in range(n):
                if mat[i][j] == 1 and row[i] == 1 and col[j] == 1:
                    count += 1

        return count
````

````python
class Solution:
    def numSpecial(self, mat: List[List[int]]) -> int:
        """
        Algorithm:
        1. Compute row count for each row
        2. For rows where row[i] == 1:
             Find the column j where mat[i][j] == 1
             Check entire column j
             If column count == 1, increment result
        3. Return result

        Time Complexity:
            O(m * n)

        Space Complexity:
            O(m)
        """

        m, n = len(mat), len(mat[0])
        row = [sum(r) for r in mat]

        count = 0

        for i in range(m):
            if row[i] == 1:
                for j in range(n):
                    if mat[i][j] == 1:
                        col_count = sum(mat[k][j] for k in range(m))
                        if col_count == 1:
                            count += 1

        return count
````
