**1550. Three Consecutive Odds**

**Easy**

**Companies**:  DJI

Given an integer array arr, return true if there are three consecutive odd numbers in the array. Otherwise, return false.
 

**Example 1:**
```python
Input: arr = [2,6,4,1]
Output: false
```
**Explanation:** There are no three consecutive odds.

**Example 2:**
```python
Input: arr = [1,2,34,3,4,5,7,23,12]
Output: true
```
**Explanation:** [5,7,23] are three consecutive odds.
 

**Constraints**:

- 1 <= arr.length <= 1000
- 1 <= arr[i] <= 1000

````python
class Solution:
    def threeConsecutiveOdds(self, arr: list[int]) -> bool:
        """
        Determines whether the array contains three consecutive odd numbers.

        Algorithm:
        1. Iterate through the array from index 0 to len(arr) - 3.
        2. For each index i:
              - Check if arr[i], arr[i+1], and arr[i+2] are all odd.
        3. If found, return True.
        4. If loop completes without finding such triplet, return False.

        Time Complexity: O(n)
            - We check each element once (at most n-2 checks).

        Space Complexity: O(1)
            - No extra space used.
        """
        n = len(arr)

        # Check every group of 3 consecutive numbers
        for i in range(n - 2):
            if arr[i] % 2 == 1 and arr[i + 1] % 2 == 1 and arr[i + 2] % 2 == 1:
                return True

        return False
````
